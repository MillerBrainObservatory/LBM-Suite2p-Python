## Environments

[UV](https://docs.astral.sh/uv/guides/install-python/#getting-started) is the default package / project manager used at the MBO.

Conda is installed as well, though `uv` is highly recommended.

See the MBO [documentation](https://millerbrainobservatory.github.io/mbo_utilities/venvs.html) on virtual-environments for more inforamtion and tips for working with `uv`.

All of the below code can be replicated on `conda` by substituting `uv pip install <PACKAGE>` with `pip install <PACKAGE>`.

MBO-Servers have pre-configured python environments.
- Located in `D://W1_SOFT//envs`
- Each environment is frozen (packages are locked) and cannot be changed in any way 
- `C:\Users\RBO\repos\mbo_utilities\.venv\Scripts`

You likely wish to maintain your own environments. This should be done in the home directory.

Here we make a `my_project` directory and create a virtual environment with `python 3.12.9`.

```ps1
PS C:\Users\MBO-Delta\my_project> uv venv --python 3.12.9
PS C:\Users\RBO\repos\mbo_utilities> .venv\Scripts\activate
(my_project) PS C:\Users\MBO-<USER>\my_project> uv pip install mbo_utilities>=2.0.0
(my_project) PS C:\Users\MBO-<USER>\my_project> uv pip install lbm_suite2p_python>=2.0.0
```

This environment can be freely added to and manipulated.

## VS Code

Most IDE's like VSCode and Pycharm will automatically detect environments when you open a folder that contains a `.venv` directory, which happens when you run `uv venv --python 3.12.9`. 

If you keep your code next to this folder, it will be easy to distinguish which environment is needed to run the code. There are two general strategies:

1. Keep all of your code and notebooks in a single place in your home directory, named projects or code or a session C://Users//MBO-User//, Cntrl + Shift + P -> Preferences: Open User Settings -> Python: Venv Path -> Enter: `D://W1_SOFT//envs`

Or, enter the path manually with Cntl + Shift + P -> Python: Select Interpreter -> Enter interpreter path -> Browse -> Navigate to `D://W1_SOFT//envs`

## Jupyter 

Jupyter is easist to set up on the terminal. You can use VSCode or Pycharms integrated terminals, or Powershell. Navigate to your project directory (with the `.venv` folder) with the `cd` command: `cd C://Users//MBO-User//projects//my_project`

Set a password for this jupyter session:

```
uv run jupyter server password
```

From within `my_project`:
```
uv run jupyter lab
```

If you previously ran `./Scripts/activate` to activate another environment, you'll see a complaint about using `--active`. That is `uv`
